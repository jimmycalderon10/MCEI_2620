```python
# ============================================================
# TALLER DE INTEGRACIÓN Y DIFERENCIACIÓN NUMÉRICA
# Diferenciación numérica de datos experimentales
# Python con NumPy
#
# Jimmy Alexander Calderon Trujillo
# Sebastian Barahona
# ============================================================

import numpy as np
import matplotlib.pyplot as plt

# ------------------------------------------------------------
# 1. CARGA DE LOS DATOS EXPERIMENTALES
# ------------------------------------------------------------

datos = np.loadtxt(
    "../../datos/datos_sensor.csv",
    delimiter=",",
    skiprows=1
)

x = datos[:, 0]
y = datos[:, 1]

N = len(x)

# ------------------------------------------------------------
# 2. VERIFICACIÓN DEL ESPACIAMIENTO
# ------------------------------------------------------------

dx = np.diff(x)
h = dx[0]

print("DIFERENCIACIÓN NUMÉRICA - PYTHON")
print("---------------------------------------------")
print(f"Número de mediciones : {N}")
print(f"Espaciado h          : {h:.1f}")
print(
    f"Espaciado uniforme   : "
    f"{'SI' if np.allclose(dx, h) else 'NO'}"
)

# ------------------------------------------------------------
# 3. DIFERENCIA HACIA ADELANTE
# ------------------------------------------------------------
#
# f'(xi) ≈ [f(xi+1) - f(xi)] / h
#
# No puede calcularse en el último punto porque no existe
# una medición posterior.
# ------------------------------------------------------------

derivada_adelante = np.full(N, np.nan)

derivada_adelante[:-1] = (
    y[1:] - y[:-1]
) / h

# ------------------------------------------------------------
# 4. DIFERENCIA CENTRAL
# ------------------------------------------------------------
#
# f'(xi) ≈ [f(xi+1) - f(xi-1)] / (2h)
#
# Solo puede calcularse en los puntos interiores.
# ------------------------------------------------------------

derivada_central = np.full(N, np.nan)

derivada_central[1:-1] = (
    y[2:] - y[:-2]
) / (2 * h)

# ------------------------------------------------------------
# 5. MOSTRAR LOS PRIMEROS RESULTADOS
# ------------------------------------------------------------

print()
print("PRIMEROS RESULTADOS")
print("-------------------------------------------------------------")
print("   x          y        Adelante       Central")
print("-------------------------------------------------------------")

for i in range(10):

    adelante = (
        "NaN"
        if np.isnan(derivada_adelante[i])
        else f"{derivada_adelante[i]:.6f}"
    )

    central = (
        "NaN"
        if np.isnan(derivada_central[i])
        else f"{derivada_central[i]:.6f}"
    )

    print(
        f"{x[i]:5.1f}   "
        f"{y[i]:10.6f}   "
        f"{adelante:>10}   "
        f"{central:>10}"
    )

# ------------------------------------------------------------
# 6. TRATAMIENTO DE LOS EXTREMOS
# ------------------------------------------------------------

print()
print("TRATAMIENTO DE LOS EXTREMOS")
print("---------------------------------------------")

print(
    f"Diferencia hacia adelante: "
    f"no se calcula en x = {x[-1]:.1f}"
)

print(
    f"Diferencia central: "
    f"no se calcula en x = {x[0]:.1f} "
    f"ni x = {x[-1]:.1f}"
)
```

```python
# ------------------------------------------------------------
# 7. GRAFICA DE LAS DERIVADAS NUMERICAS
# ------------------------------------------------------------

plt.figure(figsize=(10, 6))

plt.plot(
    x,
    derivada_adelante,
    "o-",
    label="Diferencia hacia adelante"
)

plt.plot(
    x,
    derivada_central,
    "s-",
    label="Diferencia central"
)

plt.xlabel("x")
plt.ylabel("dy/dx")
plt.title("Diferenciación numérica de los datos experimentales")
plt.grid(True)
plt.legend()
plt.tight_layout()

plt.show()
```
